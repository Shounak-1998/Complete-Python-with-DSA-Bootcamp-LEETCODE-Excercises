```python
#Task: Create three variables:
# 1.Store the word "Python" in a variable named course_name.
# 2.Store the number 25 in a variable named duration.
# 3.Store the decimal value 4.9 in a variable named rating
course_name = "Python"
duration = 25
rating = 4.9
complex = 2 + 4j

# Print the value of each variable alongside its dynamic data type using Python's built-in type() function
print("Value :", course_name, ", Type: ", type(course_name))
print("Value :", duration, ", Type: ", type(duration))
print("Value :", rating, ", Type: ", type(rating))
print("Value :", complex, ", Type: ", type(complex))
```

```python
#Part 1: Identify the Data Type

# Write the data type of each value:

#100
#99.99
#"Hello"
#False
#4 + 5j
#[10, 20, 30]
#(10, 20, 30)
#{"name": "Alice", "age": 20}
#{1, 2, 3}
#None

print(type(100))
print(type(99.99))
print(type("Hello"))
print(type(False))
print(type(4 + 5j))
print(type([10, 20, 30]))
print(type((10, 20, 30)))
print(type({"name": "Alice", "age": 20}))
print(type({1, 2, 3}))
print(type(None))
```
