```python
#Variables
#Variables are containers for storing data values

#Creating Variables
#Python has no command for declaring a variable A variable is created the moment you first assign a value to it

x=5
y="Shounak"
print(x)
print(y)
```

```python
#Variables do not need to be decalred with any particular type, and can even change type after they have been set

x = 4
x = "Shounak"

print(x)
```

```python
# Create a Calculator

num1 = float(input("Enter the first number"))
num2 = float(input("Enter the second number"))


sum = num1 + num2
sub = num1 - num2
multiply = num1 * num2
division = num1 / num2

print("Sum :", sum)
print("Sub :", sub)
print("Multiply :", multiply)
print("Division :", division)
```

```python
# Casting

#If you wwant to specify the data type of a variable, this can be done with casting

x = str(3)
y = int(3)
z = float(3)

print(x)
print(y)
print(z)
```

```python
# Get the Type
# You can get the data type of a variable with the type() function

x = 5
y = "Shounak"

print(type(x))
print(type(y))
```

```python
# Single or Double Quote

#String varaibles can be declared either by using single or double quotes

x = "Shounak"
# is the same as
x = 'Shounak'
```

```python
# Create and Assign: 
# Create a variable named user_age and assign it the integer 25. 
user_age = 25
# Create another variable named user_name and assign it the string "Alex"
user_name = "Alex"

print(user_age)
print(user_name)
```

```python
# Dynamic Typing: 
# Print the data type of both variables using the type() function.

print(type(user_age))
print(type(user_name))
```

```python
# Variable Reassignment: 
# Change the value of user_age to a float 26.5 and print its new data type

user_age = 26.5
print(user_age)
print(type(user_age))
```

```python
# Multiple Assignment: 
# In a single line of code, assign three new variables (city, country, and zip_code) with the values "Paris", "France", and 75001 respectively.
city, country, zip_code = "Paris", "France", 75001
print(city)
print(country)
print(zip_code)
```

```python
# Expression Output: 
# Create a variable called greeting. 
# Use string concatenation or an f-string to combine your variables so it prints: "Hello Alex from Paris, France!"
greeting = f"Hello {user_name} from {city}, {country}!"
print(greeting)
```

```python
# The Swapping Trick: 
# Create two variables, a = 5 and b = 10. 
# Swap their values in a single line of code without using a third temporary variable. Print a and b to verify the swap
a = 5
b = 10
a, b = b, a  # Pythonic swap
print(f"After swap -> a: {a}, b: {b}")
```
