```python
## while loop

## The while loop continuos to execute as long as the condition is True.

count=0

while count<5:
    print(count)
    count=count+1
```

```python
count=0

while count%2 == 0:
    print(count)
    count=count+1
```

```python
## Loop Control Statements

## break
## The break statements exist the loop prematurely

# break statement

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
## Continue

## The continue statement skips the current iteration and continues with the next.

for i in range(10):
    if i%2 == 0:
        continue
    print(i)
```

```python
## Pass

## The pass statement is a null operation; it does nothing.

for i in range(5):
    if i == 3:
        print("the number is 3")
        pass
    print(i)
```

```python
## Nested Loops
## A loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Examples -  Calculate the sum of first N natural numbers using a while and for loop

## while loop

n=10
sum=0
count=1

while count <= n:
    sum = sum + count
    count = count + 1
print("Sum of first 10 Natural nnumber :", sum)
```
