```python
#17. BMI Category Checker

#Formula:

#bmi = weight / (height * height)

#Categories:

#Underweight (<18.5)
#Normal (18.5–24.9)
#Overweight (25–29.9)
#Obese (30+)

is_weight= float(input("Audencia Enter your weight :"))
is_height= float(input("Audencia Enter your height :"))

bmi = is_weight / (is_height * is_height)

if bmi < 18.5:
    print("Underweight")
elif bmi >= 18.5 and bmi <=24.9:
    print("Normal")
elif bmi >= 25 and bmi <=29.9:
    print("Overweight")
elif bmi >= 30:
    print("Obese")
else:
    print("Either height or weight is missing")
```

```python
#18. Restaurant Bill System

#Conditions:

#Bill > ₹2000 → 15% discount
#Bill > ₹1000 → 10% discount
#Otherwise → No discount

#Print final amount.

is_bill = float(input("Audencia Enter a Bill Amount :"))


if is_bill < 0:
    print("Bill is not valid")
else:
    if is_bill > 2000:
        final = is_bill - (is_bill * (15 /100))
        print("You will get 15% discount :", final)
    elif is_bill > 1000:
        final = is_bill - (is_bill * (10 /100))
        print("You will get 10% discount :", final)
    else:
        print("No dicount")
```

```python
#20. Exam Result Processing

#Conditions:

#Marks in Mathematics
#Marks in Science
#Marks in English

#Rules:

#Every subject ≥ 35
#Average ≥ 60

#Output:

#Distinction
#First Class
#Pass
#Fail

is_math = int(input("Enter the Marks in Mathematics :"))
is_science = int(input("Enter the Marks in Science :"))
is_english = int(input("Enter the Marks in English :"))

is_avg = (is_math + is_science + is_english) / 3
if is_math >= 35 and is_science >=35 and is_english >= 35:
    if is_avg >=60:
        if is_avg > 75:
            print("Distinction")
        elif is_avg > 65 and is_avg <= 75:
            print("First Class")
        elif is_avg >= 60 and is_avg <= 65:
            print("Pass")
    else:
        print("Avergae is less than 60 Fail")
else:
    print("Minimum Critieria for Every Subject does not qualify")
```
