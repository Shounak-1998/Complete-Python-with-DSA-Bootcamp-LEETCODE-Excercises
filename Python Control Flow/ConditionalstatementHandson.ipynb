```python
## Nested Conditional statement
# You can place one or more if, elif, or else statements inside another if, elif, or else statement to create nested conditional statement

# To check input is positive, even or odd

num1 = int(input("Shounak Enter the Luck Number :"))

if num1>0:
    print("Oh Shounak! Number you enter is positive")
    if num1%2==0:
        print("And Shounak! Number you enter is even")
    else:
        print("And Shounak! Number you enter is odd")
        
else:
    print("And Shounak! So Number is Zero or Negative")
```

```python
## Practical Examples

## Determine if a year is a leap year using nested condition statement

year = int(input("Shounak Enter the Year to check if it is leap or not :"))

if year>0:
    print("Shounak year enter is valid you can proceed next")
    if year%4 == 0:
        if year%100 == 0:
            if year%400 == 0:
                print("Shounak year you enter is leap year")
            else:
                print("Shounak year you enter is not a leap year")
        else:
             print("Shounak year you enter is leap year")
else:
    print("Shounak year you enter is either zero or not valid or negative or does not exists")
```

```python
#5. Pass or Fail

#If marks are:

#40 or above → Pass
#Below 40 → Fail

#Example

#Enter marks: 65

#Pass

marks = float(input("Enter marks :"))

if marks >= 40:
    print("Pass")
elif marks < 40:
    print("Fail")
else:
    print("Not a valid marks")
```

```python
#7. Voting Eligibility

#Check if a person can vote.

#Conditions:

#Age ≥ 18 → Eligible
#Otherwise → Not eligible

#Example

#Age: 22

#Eligible to vote

is_age = int(input("Age :"))

if is_age ==0:
    print("Invalid Age to Vote")
elif is_age >=18:
    print("eligible to vote")
elif is_age <18:
    print("Not eligible to vote")
```

```python
#8. Number Range Checker

#Check if a number belongs to:

#1–10
#11–20
#21–30
#Outside range

#Example

#Enter number: 15

#Number is in range 11-20

is_range = int(input("Enter number :"))

if is_range >=1 and is_range<=10:
    print("Number is in range 1-10")
elif is_range >=11 and is_range<=20:
    print("Number is in range 11-20")
elif is_range >= 21 and is_range<=30:
    print("Number is in range 21-30")
elif is_range <=0 or is_range>30:
    print("Outside range")
else:
    print("Not a valid range")
```

```python
#9. Divisibility Checker

#Check whether a number is:

#Divisible by both 3 and 5
#Only by 3
#Only by 5
#Neither

#Example

#Enter number: 30

#Divisible by both 3 and 5

is_div = int(input("Enter number :"))

if is_div%3==0 and is_div%5==0:
    print("Divisible by both 3 and 5")
elif is_div%5==0:
    print("Divisible by 5")
elif is_div%3==0:
        print("Divisible by 3")
else:
    print("Neither")
```

```python
#11. ATM Withdrawal System

#Conditions:

#Balance = ₹5000
#User enters withdrawal amount

#Check:

#Amount > Balance → Insufficient balance
#Amount <= Balance
#Amount > 0 → Withdrawal successful
#Otherwise → Invalid amount

is_amount = float(input("User enters withdrawal amount"))
Balance = 5000

if is_amount > Balance:
    print("Insufficient balance")
elif is_amount <= Balance and is_amount > 0:
    print("Withdrawal successful")
else:
    print("Invalid amount")
```
